```python
x = [1,2,3,4,5]
y=[2,3,7,1,0]
```

```python
plt.plot(x,y,
         color="blue",
         linestyle="--",
         marker="o",
         markersize=10,
         markerfacecolor="red",
         label="Proyeccion de ventas")
plt.title("Ventas mensuales")
plt.xlabel("Mes")
plt.ylabel("Ventas")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#comparar resultados
y2=[3,2,8,3,1]
plt.plot(
    x,y,
         color="blue",
         linestyle="--",
         marker="o",
         markersize=5,
         markerfacecolor="red",
         label="Proyeccion de ventas"
         )
plt.plot(
    x,
    y2,
    color="black",
    label="Real venta",
    marker="s"
) 
plt.annotate("Desviacion", 
             xy=(2,2), 
             xytext=(2.5,1),
             arrowprops=dict(
                 arrowstyle="->",
                 facecolor="green")
            )
 
plt.title("Ventas mensuales")
plt.xlabel("Mes")
plt.ylabel("Ventas")
plt.legend()
plt.grid(True)
plt.show()
  
```

```python
x=np.random.rand(50)  #edad
y=np.random.rand(50)  #ingreso
z=np.random.rand(50)  #impacto

plt.scatter(
    x,
    y,
    s=z*1000,
    c="red",
    alpha=0.5, #transparencia
    label="Impacto Economico"
    #marker="s"
    )
plt.title("Impacto Economico")
plt.xlabel("Edad")
plt.ylabel("Ingreso")
plt.legend()
plt.show()
```

```python
# 100 columnas ---> df --> len(df)

colores=np.random.rand(50)

grafico_dispercion=plt.scatter(

    x,

    y,

    s=z*1000,

    c=colores,

    alpha=0.5,

    cmap='inferno'

)

plt.colorbar(grafico_dispercion,label='Intensidad')

plt.show()
```

```python
categorias  = ['A','B','C','D']
valores = [25,40,15,30]
barras = plt.bar(categorias,valores,label='Valores')
plt.title("Grafico de barras")
plt.legend()

# Etiquetas encima de cada barra
plt.bar_label(barras)
plt.show()
```
